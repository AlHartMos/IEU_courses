# 1) Motivation: Dictionaries are great… until they aren’t

### Modeling “things” with dictionaries
A dictionary can represent a structured record:

```python
student = {"name": "Ada", "email": "ada@uni.edu", "grades": [18, 15, 20]}
```

This is a very common pattern (and often used for JSON).

### The problem
As soon as you add *behavior*, you typically write many functions:

- `add_grade(student, grade)`
- `average(student)`
- `is_passing(student, threshold)`

This works, but as programs grow:
- You must remember the exact keys everywhere (`"grades"`, `"email"`, …).
- It’s easy to build inconsistent records (missing keys, wrong types).
- The code “about students” is scattered across many functions.

### The object idea (one sentence)
An **object** bundles **data + behavior** in one place:
- data → *attributes*
- behavior → *methods*


````python
# Dictionary-based approach (fine for small scripts)

# DEFINE DATA
student = {"name": "Ada", "email": "ada@uni.edu", "grades": [18, 15, 20]}

# DEFINE METHODS (FUNCTIONS) THAT ACT ON DATA
def add_grade(s, grade):
    s["grades"].append(grade)

def average(s):
    g = s["grades"]
    return sum(g) / len(g) if g else 0.0


add_grade(student, 17)
print(student)
print("Average:", average(student))
````

# 2) Class vs Object (instance)

### Definitions
- A **class** is a *blueprint* (a definition of a new type).
- An **object** (or **instance**) is a concrete value created from a class.

### Why “blueprint” is a helpful metaphor
A class describes:
- what data objects have (attributes)
- what operations they support (methods)

Then each object stores its own data.

Let’s start with the smallest class possible.


````python
class Empty:
    # HERE THERE WILL BE CODE TO DEFINE ATTRIBUTES AND METHODS
    pass

x = Empty()
y = Empty()

print("type(x) =", type(x))
print("x == y ?", x == y)     # default equality is identity (usually False for different objects)
print("x is y ?", x is y)     # identity check: are they the same object?
````

# 3) `__init__` and instance attributes

### What happens during construction?
When you write:
```python
s = Student("Ada", "ada@uni.edu")
```
Python:
1. creates a new empty Student object
2. calls `Student.__init__(s, "Ada", "ada@uni.edu")`

### What is `self`?
`self` is **the current object**.
- Inside methods, `self` is how you access “my attributes”.
- `self.name` means “the `name` attribute stored on this object”.

### Instance attributes
Attributes like `self.name` are **instance attributes**:
- each object has its own copy
- changing one student’s `name` doesn’t change another’s


# 4) Methods: functions that belong to an object

A **method** is a function defined inside a class.

Why methods are useful:
- They naturally group behavior with the data it uses.
- They provide a clean interface:
  - `ada.add_grade(18)`
  - instead of `add_grade(ada_dict, 18)`

Let’s add:
- `add_grade(grade)`
- `average()`

Notice:
- `add_grade` changes the object (side effect).
- `average` returns a computed value.


````python
class Student:
    def __init__(self, name, email):
        self.name = name
        self.email = email
        self.grades = []

    def add_grade(self, grade):
        self.grades.append(grade)

    def average(self):
        return sum(self.grades) / len(self.grades) if self.grades else 0.0

ada = Student("Ada", "ada@uni.edu")
ada.add_grade(18)
ada.add_grade(16)

print("Grades:", ada.grades)
print("Average:", ada.average())
````

# 6) Identity vs equality, and aliasing

### Identity (`is`)
`a is b` checks if both variables refer to the **same object** in memory.

### Equality (`==`)
`a == b` checks if two values are “equal”.  
For custom objects, the default `==` behaves like identity unless you define `__eq__`.

### Aliasing (common pitfall)
If you do:
```python
a = Student(...)
b = a
```
then `a` and `b` are two names for the **same object**.  
Mutating via one name affects the other.

This is the same phenomenon you saw with lists and dictionaries.


````python
class Box:
    def __init__(self):
        self.items = []

b1 = Box()
b2 = b1        # alias (not a copy)
b2.items.append("x")

print("b1.items:", b1.items)
print("b2.items:", b2.items)
print("b1 is b2:", b1 is b2)
````

# 7) Class attributes vs instance attributes

### Instance attribute
- created as `self.x = ...` (inside `__init__` or methods)
- belongs to **one object**

### Class attribute
- created directly in the class body (or assigned to the class)
- shared by **all instances**

Example:
```python
class Student:
    school = "My University"   # class attribute
```

All students see the same `school` unless overridden.

Class attributes are useful for:
- constants shared by all objects
- counters (IDs)
- configuration shared across objects


````python
class Student:
    school = "Example University"  # class attribute

    def __init__(self, name):
        self.name = name           # instance attribute

a = Student("Ada")
b = Student("Bob")
````

If you print a custom object, Python defaults to something like:

````python
# With __repr__ methods
class Person:
    def __init__(self, name: str):
        self.name = name

    def __repr__(self):
        return f"Person(name={self.name!r})"

p = Person("Ada")
print(p)
````

## Checkpoint exercise 9 — Gradebook queries

Implement:
- `class_average()` → average of student averages (0.0 if no students)
- `top_student()` → the Student with the highest average (or `None` if no students)

Tip:
- iterate over `self.students_by_email.values()`


````python
# EXERCISE 9 (TODO)

class Gradebook:
    def __init__(self):
        self.students_by_email = {}

    def add_student(self, name, email):
        if email in self.students_by_email:
            raise ValueError(f"Duplicate email: {email}")
        self.students_by_email[email] = Student(name, email)

    def add_grade(self, email, grade):
        if email not in self.students_by_email:
            raise KeyError(f"Unknown email: {email}")
        self.students_by_email[email].add_grade(grade)

    def class_average(self):
        if bool(not self.students_by_email):
            return 0.0
        else:
            grade_sums = 0
            for stu in self.students_by_email:
                student = self.students_by_email[stu]
                grade_sums += student.average()
            return grade_sums / len(self.students_by_email.keys())

    def top_student(self):
        if bool(not self.students_by_email):
            return None
        else:
            top_grade = 0
            top_student = None
            for stu in self.students_by_email:
                student = self.students_by_email[stu]
                if top_grade < student.average():
                  top_grade = student.average()
                  top_student = student
            return top_student


# TODO: test
gb = Gradebook()
gb.add_student("Ada", "ada@uni.edu")
gb.add_student("Bob", "bob@uni.edu")
gb.add_grade("ada@uni.edu", 18)
gb.add_grade("bob@uni.edu", 12)
print("Class avg:", gb.class_average())
print("Top:", gb.top_student())
````

# 10) Common pitfall: mutable default arguments

Consider:

```python
class Student:
    def __init__(self, name, grades=[]):
        self.name = name
        self.grades = grades
```

This is a bug because **default arguments are evaluated once** when the function is defined, not each time it’s called.

So all students created with the default share the **same list**.

Let’s see it.


````python
class BadStudent:
    def __init__(self, name, grades=[]):   # BUG: shared default list
        self.name = name
        self.grades = grades

a = BadStudent("Ada")
b = BadStudent("Bob")

a.grades.append(20)

print("Ada grades:", a.grades)
print("Bob grades:", b.grades, "(surprise!)")
````

## Fix pattern

Use `None` as the default, then create a new list inside:

```python
def __init__(..., grades=None):
    self.grades = [] if grades is None else list(grades)
```

This ensures each object gets its own list.
