## 1) What is flow control?

By default, Python runs your code **top to bottom**, one line after the other.

**Flow control** is everything that lets you *change that default flow*:

- **Decide** whether a block of code should run (conditionals).
- (Later) **Repeat** a block of code multiple times (loops).

In this notebook, we will learn the decision-making side: **conditions** and **`if` statements**.


### Mini-demo: code with and without a condition

````python
x = 3
print("Always runs")

if x > 0:
    print("Runs only if x is positive")

print("Always runs again")
````

## 3) Comparison operators

Comparison operators compare two values and produce a Boolean (`True` or `False`).

Common comparison operators:

- `==` : equal
- `!=` : not equal
- `<`  : less than
- `>`  : greater than
- `<=` : less than or equal
- `>=` : greater than or equal

These operators are the **foundation** of decision-making with `if`.


### Comparing numbers
Comparison operators work for integers and floats.


````python
x = 3
y = 7

print(x == y)
print(x != y)
print(x < y)
print(x >= y)
````

### Comparing strings
Strings can also be compared.

- `==` checks if two strings match exactly (case matters).
- `<` and `>` compare **lexicographically** (dictionary order).

Lexicographic order is based on character codes, so uppercase/lowercase can surprise you.


````python
print("hi" == "hi")
print("hi" == "Hi")

print("apple" < "banana") # apple comes first in the dictionary, so it is 'smaller'
print("Zebra" < "apple")  # ASCII num for capital letters is greater
````

### Membership test with `in`
For strings, you can test if a substring appears inside another string:

- `"cat" in sentence`
- `"cat" not in sentence`

This is extremely useful for conditions.


````python
sentence = "I love cats"
print("cat" in sentence)
print("dog" in sentence)
````

## 6) The `elif` statement (multiple branches)

Use `elif` when you want to check **several conditions**.

```python
if <condition1>:
    ...
elif <condition2>:
    ...
else:
    ...
```

Important rules:

- Python checks the conditions **from top to bottom**.
- The **first** condition that is `True` wins.
- If no condition is `True`, the `else` block runs (if present).

This means that the **order of your `elif` clauses matters**.


````python
x = -5

if x > 0:
    print("x is positive")
elif x < 0:
    print("x is negative")
else:
    print("x is zero")
````

````python
# Format a number in two different ways
num = 99
out = f"{num:.2f}" if num < 100 else f"{num:.0f}" # if the number is smaller than 100, format the number as a float with 2 decimal points
print(out)

num = 123
out = f"{num:.2f}" if num < 100 else f"{num:.0f}"
print(out)
````

## 9) Logical operators: `and`, `or`, `not`

Logical operators let you combine conditions.

### `and`
`A and B` is `True` only if **both** `A` and `B` are `True`.

### `or`
`A or B` is `True` if **at least one** of them is `True`.

### `not`
`not A` flips a Boolean: `True` becomes `False` and vice-versa.

Tip: Use parentheses for readability in complex conditions.


````python
# Pass rule: grade >= 5 AND attendance >= 80
grade = 8
attendance = 89

if (grade >= 5) and (attendance >= 80):
    print("You passed")
else:
    print("You have not passed")
````

````python
# Fail rule: grade < 5 OR attendance < 80
grade = 8
attendance = 60

if (grade < 5) or (attendance < 80):
    print("You have not passed")
else:
    print("You passed")
````

````python
# Add a dot at the end of a string if it doesn't have one
s = "Hello world"

if not s.endswith("."):
    s = s + "."

print(s)
````

````python
# Exclusive or

def xor(a, b):
  if(a and b) or (not a and not b):
    # a == b
    return False
  else:
    # (a or b) and (not (a and b))
    return True

print(xor(True, True))
print(xor(True, False))
print(xor(False, True))
print(xor(False, False))
````

### Short-circuit logic (very important!)

Python evaluates `and` / `or` from left to right.

- For `A and B`: if `A` is `False`, Python **does not evaluate** `B`.
- For `A or B`: if `A` is `True`, Python **does not evaluate** `B`.

This can prevent crashes.


````python
x = 0

# Safe division: the second part is evaluated only if x != 0
if (x != 0) and (10 / x > 2):
    print("big")
else:
    print("safe")
````

## 10) Debugging flow control

Conditionals often go wrong in a few predictable ways:

- **Indentation errors**: code is inside/outside the block by mistake.
- **`=` vs `==`**: `=` assigns, `==` compares.
- **Wrong order** in `elif`: a general case placed before a special case.
- **Wrong variable type**: comparing strings vs numbers.

A helpful technique is to print intermediate values, or use `assert` to check assumptions.

`assert <condition>` will crash the program if the condition is `False`.

### Advanced Challenge 1 — Triangle classifier

Given three side lengths `a`, `b`, `c`, write a program that:

1) Prints `"invalid"` if the sides cannot form a triangle.

A triangle is invalid if:
- at least one side is non-positive, or
- the triangle inequality fails: each side must be **strictly smaller** than the sum of the other two.

2) Otherwise prints the type:
- `"equilateral"`: all sides equal
- `"isosceles"`: two sides equal
- `"scalene"`: all sides different

3) (Bonus) If it is a **right triangle**, append `" + right"`.



````python
# ✍️ Your turn
a, b, c = 3, 4, 5

# TODO: write your classifier
def valid_triangle(a, b, c):
  if (a <= 0) or (b <= 0) or (c <= 0):
    return False
  elif (a >= b + c) or (b >= a + c) or (c >= a + b):
    return False
  else:
    return True

a, b, c = 3, 4, 5

if (not valid_triangle(a, b, c)):
  print("Invalid")
elif (a == b == c):
  print("Equilateral")
elif (a == b) or (b == c) or (a == c):
  print("Isosceles")
else:
  print("Scalene")

if (a**2 + b**2 == c**2) or (a**2 + c**2 == b**2) or (c**2 + b**2 == a**2):
  print(" + right")
````
