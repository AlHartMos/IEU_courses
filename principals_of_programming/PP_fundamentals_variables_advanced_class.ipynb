## Variables in Python

In computer programming, **variables** allow us to store and reuse data.

### Definition: variable
A **variable** is a *name* that is **bound** to a **value** (an object) in memory.

In Python, the assignment operator `=` binds a name (left) to a value (right).

Example:


````python
x = 4
y = 9

print(x)
print(y)
````

### Variables inside expressions
Once a variable is defined, you can use it inside computations (expressions):


````python
x = 4
y = 9

print(x + y)
print(x * y)
print(y - x)
````

### Reassignment (rebinding)
You can **reassign** a variable: bind the name to a new value.

Important idea: Python does **not** “change the past”.
It simply changes what the name points to **from now on**.


````python
x = 4
print(x)

x = 10  # rebinding
print(x)

x = x + 1
print(x)
````

### Augmented assignment (`+=`, `-=`, `*=`, ...)
Python offers a compact way to update a variable using its current value.

Example:
- `x += 1` means “take the current value of `x` and add 1”
- It is equivalent to `x = x + 1`


````python
x = 7
x += 1
print(x)

x *= 3
print(x)
````

### Dynamic typing (powerful, but you must stay aware)
Python is **dynamically typed**:
- you do **not** declare variable types (`int`, `float`, `str`, ...)
- a name can be rebound to a value of a different type later

You can always inspect the type with `type(...)`.


````python
x = 10
print(type(x))

x = 10.0
print(type(x))

x = "ten"
print(type(x))
````

````python
# Define your variables before using them!!!!
x = 10
z = x * x
print(x + z)
````

### Immutability: why “updating” numbers and strings creates new values
Some Python types are **immutable**: once created, they cannot be changed *in place*.

Integers (`int`) and strings (`str`) are immutable.
This is why “updating” them means *creating a new value* and rebinding the variable name.

We can *peek* at identity using `id(...)` (an internal identifier):

**Why does Python do this?**

Immutability has important advantages:

1. Safety: If objects can’t change, they can’t be accidentally modified elsewhere.

2. Predictability: When you see a number or a string, you know it won’t suddenly change.

3. Efficiency: Python can safely reuse immutable objects internally.


````python
x = 10.0
print("x =", x, "id(x) =", id(x))

x = x + 1
print("x =", x, "id(x) =", id(x))
````

### Integers vs floats
- `int` represents integers (…, -2, -1, 0, 1, 2, …)
- `float` represents real numbers *approximately* (e.g., `3.14`)

Division `/` always returns a `float`, even if the result looks like an integer:


````python
print(8 / 4)
print(type(8 / 4))

print(8 // 4)
print(type(8 // 4))
````

### Operator precedence (order of operations)
Python follows the usual math rules:

1. Parentheses `(...)`
2. Exponents `**`
3. Multiplication/division `*`, `/`, `//`, `%` (left to right)
4. Addition/subtraction `+`, `-` (left to right)

When in doubt: **add parentheses**. Readability beats cleverness.


````python
print(2 + 3 * 4)      # multiplication first
print((2 + 3) * 4)      # parentheses first
print(2 ** 3 ** 2)      # powers associate to the right: 2 ** (3 ** 2)
````

## String Methods

Strings are not just simple variables: they are **objects**.

### Definition: object and method
- An **object** is a value with both **data** and **behavior**.
- A **method** is a function *attached to an object* and called with dot notation:

`some_string.method_name(...)`

Most string methods return a **new string** (because strings are immutable).


### Case conversion
Common methods:
- `.upper()` → all uppercase
- `.lower()` → all lowercase
- `.title()` → title case (each word capitalized)
- `.capitalize()` → capitalize first character


````python
s = "Hello world"

print(s.upper())       # ALL CAPS
print(s.lower())       # all lowercase
print(s.capitalize())  # only first character uppercase
print(s.title())       # Title Case (each word capitalized)
print(s.swapcase())    # swap lower ↔ upper

# Notice: the original string didn't change (strings are immutable)
print("original:", s)
print("upper: ", s.upper())
print("lower: ", s.lower())
print("title: ", s.title())
print("swapcase: ", s.swapcase())
````

### Accessing and slicing strings in Python

In Python, a string is an **ordered sequence of characters**, meaning that each character has a specific position (index). You can access individual characters or extract parts of a string using **indexing** and **slicing**.

**Indexing** allows you to access a single character using square brackets `[]`. Indexing starts at `0`, so `s[0]` refers to the first character. Python also supports **negative indexing**, where `-1` refers to the last character.

**Slicing** allows you to extract a substring using the syntax `s[start:stop]`. The slice includes the character at index `start` but excludes the character at index `stop`. Either index can be omitted to mean “from the beginning” or “to the end”. You can also include a **step** to skip characters or reverse the string.

Although you can access and slice strings, **strings are immutable**: once created, their characters cannot be changed in place. Any operation that seems to “modify” a string actually creates a new one.



````python
s = "Python programming"

# Accessing individual characters
print(s[0])      # first character
print(s[1])      # second character
print(s[-1])     # last character
print(s[-2])     # second-to-last character
````

````python
# Slicing substrings
print(s[0:6])    # 'Python'
print(s[7:18])   # 'programming'
print(s[:6])     # from start
print(s[7:])     # to end
````

````python
# Slicing with a step
print(s[::2])    # every second character

# Reversing a string
print(s[::-1])

# Strings are immutable
# The following line would raise an error:
# s[0] = "J"

# Creating a new string instead
new_s = "J" + s[1:]
print(new_s)
````

### String formatting helpers: `strip`, `lstrip`, `rstrip`, `center`

### Chaining methods (and why order matters)
Because methods often return new strings, you can chain them:

`text.replace(...).upper().strip()`

Read chained code left-to-right like a pipeline.

⚠️ **Order matters.** If you call `.upper()` first, the text changes — and a later `.replace("false", "true")`
might no longer match (because `"false"` becomes `"FALSE"`).


````python
s = "  This statement is false.  "

result = s.replace("false", "true").upper().strip()
print(result)
````

````python
s = "This statement is false."

# Order matters:
print(s.upper().replace("false", "true"))   # "false" won't match after uppercasing
print(s.replace("false", "true").upper())   # works as intended
````

### Definition: debugging
**Debugging** is the process of identifying and fixing errors in our code.

There are three common kinds of problems:
1. **Syntax errors**: Python cannot *parse* the code (missing quotes, parentheses, etc.)
2. **Runtime errors**: code runs until it crashes (e.g., `NameError`, `TypeError`)
3. **Logic errors**: code runs, but the result is wrong

A good habit: read the error message carefully — it usually tells you exactly where to look.


### Reading error messages (tracebacks)
A traceback typically contains:
- the file/cell and line number
- the line of code that failed
- the error type and message (e.g., `SyntaxError`, `NameError`)

Start at the **bottom line** of the traceback: that is the actual error.

### `assert`: a debugging checkpoint

The `assert` statement is a simple but powerful tool used to **check assumptions** while a program is running.

The basic syntax is:
```python
assert condition



````python
# Checking whether a variable satisfies a condition
x = 10
x = x/10
x = x-x

assert x > 0, "x must be positive"
````

````python
# Assert can be combined with print
x = 0.5
y = ((3 * x - 1) / (x ** 3)) ** (1/2)

assert y == 2.0
print("The assertion passed; y is correct.")
````

````python
# Multiple asserts can be used
temperature = 1000.0
assert temperature > 0
assert temperature < 100
````
