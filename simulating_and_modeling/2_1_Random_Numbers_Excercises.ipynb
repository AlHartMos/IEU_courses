#Excercise 2

Implement the Linear Congruential Method in a function and test it with different parameters. Feel free to copy the function from our class

```python
#Function
def LCM (n, seed, a, c, m):
  pseudo_random_list = []
  pseudo_random_list.append(seed)
  for i in range(1, n + 1):
    pseudo_random_list.append((a * pseudo_random_list[i - 1] + c) % m)
  uniform = np.array(pseudo_random_list) / m
  return(list(uniform))
```
