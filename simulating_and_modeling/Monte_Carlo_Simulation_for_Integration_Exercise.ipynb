# Monte Carlo Simulation for Integration Exercise

Suppose we want to estimate the integral of a function over a certain interval. Specifically, let's estimate the integral of $f(x) = x^2$ over the interval `[0, 1]`. Analytically, we know the exact value of this integral is $\frac{1}{3}$, but we'll estimate it using Monte Carlo simulation to illustrate the method.

**Approach Using Monte Carlo Simulation:**

1. Random Point Generation: Generate a large number of random points
$x$ within the interval `[0, 1]`.
2. Evaluate Function at Each Point: For each random point $x$, evaluate $f(x) = x^2$.
3. Estimate Integral: The average of these function values, multiplied by the length of the interval (which is 1 in this case), gives an estimate of the integral.

Next we see a visual representation of what we have to simulate:

```python
import matplotlib.pyplot as plt
import numpy as np

x_curve = np.linspace(0, 1, 100)
y_curve = x_curve**2

num_random_points_for_plot = 100
x_random = np.random.rand(num_random_points_for_plot)
y_random = x_random**2

plt.figure(figsize=(8, 6))
plt.plot(x_curve, y_curve, label="f(x) = x^2", color='blue')
plt.scatter(x_random, y_random, color='red', alpha=0.5)
plt.fill_between(x_curve, y_curve, color='blue', alpha=0.3)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("f(x) = x^2 estimation in the interval [0, 1]")
plt.legend()
plt.grid(True)
plt.show()
```


**Breakdown of the Process:**

1. Random Point Generation: generate 10,000 random points within the interval [0, 1]. Each point represents a value of $x$.
2. Evaluating the Function: For each randomly generated point $x$, we calculate $f(x) = x^2$.
3. Estimating the Integral: We obtain the average of the resulting values for all the random points and multiply them by the length of the interval. In this case, the interval length is 1, so the estimated integral is simply the average of the function values.

**Expected output** (`random.random()`): `0.3332677026933688`

**Expected output** (`np.random.uniform(0,1,1)`): `0.33321910246160047`

```python
import random
import numpy as np

np.random.seed(2026)

# INSTRUCTIONS:
  # Number of replicas of the experiment
  # Sum of all estimated integrals from each replica
  # Number of random points to generate
  # Function to integrate
  # Replicate the experiment 1000 times
  # Calculate the average of all estimated integrals

# Integrate x squared using the monte carlo method and repeat the experiment [num_trials] amount of times
def integrate_x_squared(num_trials, num_points = 10_000):
  # Array to hold results
  results = []

  # Loop over experiment for the specified amount of trials
  for i in range(num_trials):
    x_points = np.random.rand(num_points) # Generates [num_points] numbers between 0 and 1
    y_points = x_points ** 2 # Generates the y points for the randomly generated x points
    integrals = y_points * 1 # Integral is equal to the y coordinate height times the length of the interval [1]
    estimated_integral = sum(integrals) / num_points # Average of all integral values
    results.append(estimated_integral) # Append the estimated value to the results

  return results

# Number of trials
num_trials = 1_000

# Integral
integral_result = sum(integrate_x_squared(num_trials)) / num_trials
print (integral_result)


```
